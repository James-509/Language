```python
import torch
from torchvision.datasets import MNIST
from torchvision import transforms

transform = transforms.Compose([
    transforms.Resize((64, 64)),
    transforms.ToTensor()
])

mnist = MNIST(
    root="data",
    train=True,
    download=True,
    transform=transform
)
```

```python
arabic_images = []
arabic_labels = []

for image, digit in mnist:
    arabic_images.append(image)
    arabic_labels.append(1)

arabic_images = torch.stack(arabic_images)
arabic_labels = torch.tensor(arabic_labels)

print(arabic_images.shape)
```

```python
import pandas as pd

df = pd.read_csv(r"C:\Users\james\Downloads\archive (2)\chineseMNIST.csv")

chinese_images = torch.tensor(
    df.iloc[:, :4096].values,
    dtype=torch.float32
)

chinese_images /= 255.0

chinese_images = chinese_images.reshape(
    -1, 1, 64, 64
)

chinese_labels = -torch.ones(len(chinese_images))
```

```python
print(chinese_images.shape)
```

```python
images = torch.cat(
    [arabic_images, chinese_images],
    dim=0
)

labels = torch.cat(
    [arabic_labels, chinese_labels],
    dim=0
)
```

```python
from torch.utils.data import TensorDataset, DataLoader

dataset = TensorDataset(images, labels)

loader = DataLoader(
    dataset,
    batch_size=64,
    shuffle=True
)
```
